```python
jobs = df[["title", "skills", "seniority"]].copy()

jobs.head()
```

```python
jobs["title"] = jobs["title"].str.lower().str.strip()
jobs["skills"] = jobs["skills"].str.lower().str.strip()
jobs["seniority"] = jobs["seniority"].str.lower().str.strip()

jobs.head()
```

```python
jobs['skills_text'] = jobs['skills'].str.replace("|"," ",regex=False).str.replace(r"\s+"," ",regex=True).str.strip()
```

* 1st This replaces each pipe symbol | with a space. regex=False means It tells pandas to treat | as a literal pipe character, not as a regular-expression pattern.

* 2nd This replaces multiple consecutive whitespace characters with a single space. regex=True tells pandas to interpret the pattern as a regular expression.

* 3rd This removes whitespace from the beginning and end of the text.

```python
jobs["job_text"] = (
    jobs["title"].fillna("") + " " +
    jobs["skills_text"].fillna("")
).str.strip()

jobs[["title", "skills_text", "job_text"]].head(10)
```

```python
print("Missing job titles:", jobs["title"].isnull().sum())
print("Missing skill text:", jobs["skills_text"].isnull().sum())
print("Missing job text:", jobs["job_text"].isnull().sum())

print(
    "Empty job text:",
    jobs["job_text"].str.strip().eq("").sum()
)
```

```python
jobs_without_skills = jobs[
    jobs["skills_text"].isna() |
    jobs["skills_text"].str.strip().eq("")
]

print("Jobs without skills:", len(jobs_without_skills))

jobs_without_skills[["title", "skills_text", "seniority"]].head(10)
```

```python
print("Duplicate job text:", jobs["job_text"].duplicated().sum())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
tfidf = TfidfVectorizer()

job_vectors = tfidf.fit_transform(
    jobs['job_text'].fillna("")
)

print(job_vectors.shape)
```

```python
print(tfidf.get_feature_names_out())
```

```python
resume_text = """
Python developer with experience in Python,
machine learning, deep learning, TensorFlow,
PyTorch, SQL, and data analysis.
"""

resume_text = resume_text.lower().strip()
```

```python
resume_vector = tfidf.transform([resume_text])
resume_vector.shape
```

```python
similarity_score = cosine_similarity(resume_vector,job_vectors).flatten()

len(similarity_score)
```

```python
top_indices = similarity_score.argsort()[::-1][:5]
recommendation = jobs.iloc[top_indices].copy()
recommendation['similarity_score'] = (
    similarity_score[top_indices]
)
recommendation[
    ['title','seniority','skills_text','similarity_score']
]
```

```python
recommendation[
    ["title", "seniority", "skills_text", "similarity_score"]
].sort_values(
    by="similarity_score",
    ascending=False
).reset_index(drop=True)
```

```python
print("Recommended job titles:")
print(recommendation["title"].value_counts())
```

```python
resume_skills = {
    "python",
    "machine learning",
    "deep learning",
    "tensorflow",
    "pytorch",
    "sql",
    "data analysis"
}

for _,row in recommendation.iterrows():
    job_skills = set(row['skills_text'].split())
    matched_skills = resume_skills.intersection(job_skills)

    print("job : ",row['title'])
    print("similarity_score : ",round(row['similarity_score'],2))
    print("matched skills : ",matched_skills)
    print("\n")
```

```python
resume_skills = {
    "python",
    "machine learning",
    "deep learning",
    "tensorflow",
    "pytorch",
    "sql",
    "data analysis"
}

comparison_results = []

for idx, row in recommendation.iterrows():

    job_skills = {
        skill.strip().lower()
        for skill in str(row["skills"]).split("|")
        if skill.strip()
    }

    matched_skills = resume_skills.intersection(job_skills)

    skill_match_percentage = (
        len(matched_skills) / len(job_skills) * 100
        if job_skills else 0
    )

    comparison_results.append({
        "title": row["title"],
        "seniority": row["seniority"],
        "similarity_score": row["similarity_score"],
        "matched_skills": ", ".join(sorted(matched_skills)),
        "skill_match_percentage": round(skill_match_percentage, 2)
    })

comparison_df = pd.DataFrame(comparison_results)

comparison_df
```

```python
comparison_df = comparison_df.sort_values(
    by="similarity_score",
    ascending=False
).reset_index(drop=True)

comparison_df
```

```python
target_seniority = "entry-level"

comparison_df["target_seniority_match"] = (
    comparison_df["seniority"] == target_seniority
)

comparison_df[
    ["title", "seniority", "target_seniority_match"]
]
```

```python
skill_match_ranking = comparison_df.sort_values(
    by="skill_match_percentage",
    ascending=False
).reset_index(drop=True)

skill_match_ranking[
    [
        "title",
        "seniority",
        "matched_skills",
        "skill_match_percentage",
        "similarity_score"
    ]
]
```

```python
comparison_df["combined_score"] = (
    0.5 * comparison_df["similarity_score"] +
    0.5 * (comparison_df["skill_match_percentage"] / 100)
)

comparison_df[
    [
        "title",
        "similarity_score",
        "skill_match_percentage",
        "combined_score"
    ]
].sort_values(
    by="combined_score",
    ascending=False
).reset_index(drop=True)
```

### Why do we calculate a combined score. Your recommendation system currently uses two measurements.
     A. Similarity score
        Calculated using TF-IDF and cosine similarity.
        It measures how similar the resume text is to the job's title and skills.
     B. Skill match percentage
        It measures how many of the job's required skills are present in the resume's skill set.

```python
resume_text_2 = """
Data analyst with experience in SQL, Python,
data analysis, statistics, Excel, Power BI,
Tableau, and data visualization.
"""

resume_text_2 = resume_text_2.lower().strip()
```

```python
resume_vector_2 = tfidf.transform([resume_text_2])
```

```python
similarity_score_2 = cosine_similarity(
    resume_vector_2,
    job_vectors
).flatten()

top_indices_2 = similarity_score_2.argsort()[::-1][:5]

recommendation_2 = jobs.iloc[top_indices_2].copy()

recommendation_2["similarity_score"] = similarity_score_2[top_indices_2]

recommendation_2[
    ["title", "seniority", "skills_text", "similarity_score"]
]
```

```python
resume_skills_2 = {
    "sql",
    "python",
    "data analysis",
    "statistics",
    "excel",
    "power bi",
    "tableau",
    "data visualization"
}

job_skill_match = []
combined_scores_2 = []

for idx, row in recommendation_2.iterrows():

    job_skills = {
        skill.strip().lower()
        for skill in str(row["skills"]).split("|")
        if skill.strip()
    }

    matched_skills = resume_skills_2.intersection(job_skills)

    skill_match_percentage = (
        len(matched_skills) / len(job_skills) * 100
        if job_skills else 0
    )

    combined_score = (
        0.5 * row["similarity_score"]
        + 0.5 * (skill_match_percentage / 100)
    )

    job_skill_match.append({
        "matched_skills": ", ".join(sorted(matched_skills)),
        "skill_match_percentage": round(skill_match_percentage, 2)
    })

    combined_scores_2.append(combined_score)

skill_match_df_2 = pd.DataFrame(job_skill_match, index=recommendation_2.index)

recommendation_2["matched_skills"] = skill_match_df_2["matched_skills"]
recommendation_2["skill_match_percentage"] = (
    skill_match_df_2["skill_match_percentage"]
)
recommendation_2["combined_score"] = combined_scores_2

recommendation_2 = recommendation_2.sort_values(
    by="combined_score",
    ascending=False
)

recommendation_2[
    [
        "title",
        "seniority",
        "similarity_score",
        "matched_skills",
        "skill_match_percentage",
        "combined_score"
    ]
]
```

```python
resume_skills_2 = {
    "sql",
    "python",
    "data analysis",
    "statistics",
    "excel",
    "power bi",
    "tableau",
    "data visualization"
}

all_job_results = []

for idx, row in jobs.iterrows():

    job_skills = {
        skill.strip().lower()
        for skill in str(row["skills"]).split("|")
        if skill.strip()
    }

    matched_skills = resume_skills_2.intersection(job_skills)

    skill_match_percentage = (
        len(matched_skills) / len(job_skills) * 100
        if job_skills else 0
    )

    similarity = similarity_score_2[idx]

    combined_score = (
        0.5 * similarity
        + 0.5 * (skill_match_percentage / 100)
    )

    all_job_results.append({
        "job_index": idx,
        "title": row["title"],
        "seniority": row["seniority"],
        "similarity_score": similarity,
        "matched_skills": ", ".join(sorted(matched_skills)),
        "skill_match_percentage": round(skill_match_percentage, 2),
        "combined_score": combined_score
    })

all_jobs_scored = pd.DataFrame(all_job_results)

all_jobs_scored.head()
```

```python
final_recommendations_2 = all_jobs_scored.sort_values(
    by="combined_score",
    ascending=False
).head(5)

final_recommendations_2.reset_index(drop=True)
```
